```python
!pip install -q kagglehub

import os, cv2, numpy as np, matplotlib.pyplot as plt, seaborn as sns
import tensorflow as tf, kagglehub
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetB0
from sklearn.metrics import classification_report, confusion_matrix

gpus = tf.config.list_physical_devices("GPU")
print("GPU:", gpus)
if gpus:
    EP1, EP2 = 10, 15
else:
    print("No GPU found: using fewer epochs. For full results enable Runtime > Change runtime type > T4 GPU")
    EP1, EP2 = 2, 2

# 1. Download dataset
path = kagglehub.dataset_download("masoudnickparvar/brain-tumor-mri-dataset")
train_dir = os.path.join(path, "Training")
test_dir = os.path.join(path, "Testing")

# 2. Load data
IMG_SIZE, BATCH = (224, 224), 32
train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir, validation_split=0.15, subset="training", seed=42,
    image_size=IMG_SIZE, batch_size=BATCH, label_mode="categorical")
val_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir, validation_split=0.15, subset="validation", seed=42,
    image_size=IMG_SIZE, batch_size=BATCH, label_mode="categorical")
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir, image_size=IMG_SIZE, batch_size=BATCH,
    label_mode="categorical", shuffle=False)
class_names = train_ds.class_names
print("Classes:", class_names)

# 3. Augmentation
augment = tf.keras.Sequential([
    layers.RandomFlip("horizontal"), layers.RandomRotation(0.08),
    layers.RandomZoom(0.1), layers.RandomContrast(0.1)])
AUTO = tf.data.AUTOTUNE
train_ds = train_ds.map(lambda x, y: (augment(x, training=True), y),
                        num_parallel_calls=AUTO).prefetch(AUTO)
val_ds = val_ds.prefetch(AUTO)
test_ds = test_ds.prefetch(AUTO)

# 4. Model
base = EfficientNetB0(include_top=False, weights="imagenet", input_shape=IMG_SIZE + (3,))
base.trainable = False
x = layers.GlobalAveragePooling2D()(base.output)
x = layers.Dropout(0.3)(x)
out = layers.Dense(len(class_names), activation="softmax")(x)
model = Model(base.input, out)
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
              loss="categorical_crossentropy",
              metrics=["accuracy"])

cb = [tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=4, restore_best_weights=True),
      tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.3, patience=2)]

# 5. Train: phase 1 (frozen base), phase 2 (fine-tune)
print("\nPhase 1: training top layers")
h1 = model.fit(train_ds, validation_data=val_ds, epochs=EP1, callbacks=cb)

print("\nPhase 2: fine-tuning")
base.trainable = True
for l in base.layers[:-40]:
    l.trainable = False
for l in base.layers:
    if isinstance(l, layers.BatchNormalization):
        l.trainable = False
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
              loss="categorical_crossentropy",
              metrics=["accuracy"])
h2 = model.fit(train_ds, validation_data=val_ds, epochs=EP2, callbacks=cb)

# 6. Training curves
plt.figure(figsize=(12, 4))
for i, m in enumerate(["accuracy", "loss"]):
    plt.subplot(1, 2, i + 1)
    plt.plot(h1.history[m] + h2.history[m], label="train")
    plt.plot(h1.history["val_" + m] + h2.history["val_" + m], label="val")
    plt.title(m); plt.legend()
plt.show()

# 7. Evaluate on test set
test_loss, test_acc = model.evaluate(test_ds)
print(f"\nTest accuracy: {test_acc:.4f}")
y_true = np.concatenate([np.argmax(y.numpy(), axis=1) for _, y in test_ds])
y_pred = np.argmax(model.predict(test_ds), axis=1)
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))
plt.figure(figsize=(6, 5))
sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Confusion Matrix")
plt.show()

# 8. Grad-CAM
grad_model = Model(model.inputs, [model.get_layer("top_activation").output, model.output])

def gradcam(img):
    img = tf.convert_to_tensor(img[np.newaxis], dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv, preds = grad_model(img)
        idx = tf.argmax(preds[0])
        loss = preds[:, idx]
    grads = tape.gradient(loss, conv)
    pooled = tf.reduce_mean(grads, axis=(0, 1, 2))
    hm = tf.squeeze(conv[0] @ pooled[..., tf.newaxis])
    hm = tf.maximum(hm, 0) / (tf.reduce_max(hm) + 1e-8)
    return hm.numpy(), int(idx), float(preds[0][idx])

for images, labels in test_ds.take(1):
    n = 6
    plt.figure(figsize=(14, 5))
    for i in range(n):
        img = images[i].numpy().astype("uint8")
        hm, pred, conf = gradcam(images[i].numpy())
        hm = cv2.resize(hm, IMG_SIZE)
        col = cv2.cvtColor(cv2.applyColorMap(np.uint8(255 * hm), cv2.COLORMAP_JET), cv2.COLOR_BGR2RGB)
        plt.subplot(2, n, i + 1); plt.imshow(img); plt.axis("off")
        plt.title("True: " + class_names[np.argmax(labels[i])], fontsize=9)
        plt.subplot(2, n, n + i + 1); plt.imshow(cv2.addWeighted(img, 0.6, col, 0.4, 0)); plt.axis("off")
        plt.title(f"{class_names[pred]} ({conf:.0%})", fontsize=9)
    plt.tight_layout(); plt.show()

# 9. Save and download
model.save("brain_tumor_model.keras")
from google.colab import files
files.download("brain_tumor_model.keras")
```
